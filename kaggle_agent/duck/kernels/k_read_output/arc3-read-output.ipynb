```python

import os,glob,re,subprocess
print(subprocess.run("find /kaggle/input -maxdepth 4 -type d | head -30",shell=True,capture_output=True,text=True).stdout)
for name in ["arc3-duck-official","arc3-duck-handbook"]:
    roots=[d for d in glob.glob("/kaggle/input/**/", recursive=True) if name in d]
    base=[r for r in roots if os.path.exists(os.path.join(r,"summary.txt"))]
    print("\n"+"#"*20, name, "root=", base[:1])
    if not base: continue
    b=base[0]
    print(open(os.path.join(b,"summary.txt")).read()[:3000])
    # 逐局
    for f in sorted(glob.glob(os.path.join(b,"*_p0.log"))):
        t=open(f).read(); m=re.findall(r"\[finished\][^\n]*",t)
        print(os.path.basename(f), m[-1][:160] if m else "(no finished line)")
    # 手册是否在提示词里
    hits=0
    for f in glob.glob(os.path.join(b,"**/*"),recursive=True):
        if os.path.isfile(f) and f.endswith((".txt",".log",".json")):
            try:
                if "Field guide" in open(f,errors="ignore").read(): hits+=1
            except Exception: pass
    print("含 'Field guide' 的文件数:",hits)
    for f in glob.glob(os.path.join(b,"*.log")):
        if "_p0" in f: continue
        t=open(f,errors="ignore").read(); print("== 主日志",os.path.basename(f),len(t)); print(t[-2500:])
```
